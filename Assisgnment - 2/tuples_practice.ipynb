# Tuple Indexing

```python
colors = ("red", "green", "blue", "yellow", "purple")

second = colors[1]
last = colors[-1]

mid = len(colors) // 2
middle_three = colors[mid - 1:mid + 2]

print("Second:", second)
print("Last:", last)
print("Middle three:", middle_three)
```

# Tuple Unpacking

```python
employee = ("Ali", 28, "Data Engineer", 150000)

name, age, designation, salary = employee

print("Name:", name)
print("Salary:", salary)
```

# Tuple Concatenation

```python
tuple1 = (10, 20, 30)
tuple2 = (40, 50, 60)

combined = tuple1 + tuple2

reversed_tuple = combined[::-1]

print("Combined:", combined)
print("Reversed:", reversed_tuple)
```

# Count and Locate

```python
numbers = (10, 20, 30, 20, 40, 20, 50)

count_20 = numbers.count(20)

first_index = numbers.index(20)

print("Count of 20:", count_20)
print("First index of 20:", first_index)
```

# Extended Tuple Unpacking

```python
data = (100, 200, 300, 400, 500, 600)

first, second, *remaining, last = data

print("First:", first)
print("Second:", second)
print("Last:", last)
print("Remaining:", remaining)
```

# Modify an Immutable Tuple

```python
numbers = (10, 20, 30, 40, 50)

idx = numbers.index(30)
new_numbers = numbers[:idx] + (300,) + numbers[idx + 1:]

print("Original:", numbers)
print("New tuple:", new_numbers)
```

# Nested Tuple Access

```python
students = (
    ("Ali", (85, 90, 88)),
    ("Sara", (92, 87, 95)),
    ("Ahmed", (78, 84, 80))
)

sara_second = students[1][1][1]

ali_scores = students[0][1]
ali_average = sum(ali_scores) / len(ali_scores)

ahmed_highest = max(students[2][1])

print("Sara second score:", sara_second)
print("Ali average:", round(ali_average, 2))
print("Ahmed highest score:", ahmed_highest)
```

# Tuple Sorting

```python
from operator import itemgetter

employees = (
    ("Ali", 80000),
    ("Sara", 120000),
    ("Ahmed", 95000)
)

sorted_employees = tuple(sorted(employees, key=itemgetter(1), reverse=True))

print("Sorted by salary (descending):", sorted_employees)
```

# Tuple Matrix

```python
import math

matrix = ((2, 4, 6), (8, 10, 12), (14, 16, 18))

diagonal = (matrix[0][0], matrix[1][1], matrix[2][2])

diagonal_product = math.prod(diagonal)

row_sums = (sum(matrix[0]), sum(matrix[1]), sum(matrix[2]))

print("Diagonal:", diagonal)
print("Diagonal product:", diagonal_product)
print("Row sums:", row_sums)
```

# Inventory Tuple Challenge

```python
from operator import itemgetter

records = (
    ("Laptop", 120000, 5),
    ("Mouse", 1500, 20),
    ("Keyboard", 3500, 10)
)

inventory_value = (records[0][1] * records[0][2]
                   + records[1][1] * records[1][2]
                   + records[2][1] * records[2][2])

most_expensive = max(records, key=itemgetter(1))[0]

names = tuple(zip(*records))[0]
reversed_names = names[::-1]

print("Total inventory value:", inventory_value)
print("Most expensive product:", most_expensive)
print("Reversed names:", reversed_names)
```
