## Accessing List Elements

```python
numbers = [10, 20, 30, 40, 50, 60, 70]

first = numbers[0]
middle = numbers[len(numbers) // 2]
last = numbers[-1]

print("First, Middle, Last:", first, middle, last)
```

# List Slicing

```python
numbers = [5, 10, 15, 20, 25, 30, 35, 40, 45, 50]


first_five = numbers[:5]


last_four = numbers[-4:]


every_second_reverse = numbers[::-2]

print("First five:", first_five)
print("Last four:", last_four)
print("Every second in reverse:", every_second_reverse)
```

# List Modification

```python
fruits = ["apple", "banana", "cherry", "mango"]


fruits[1] = "orange"


fruits.insert(2, "grapes")


fruits.remove("cherry")

print("Updated fruits list:", fruits)
```

# Reverse Without reverse

```python
numbers = [12, 45, 23, 67, 89, 34]


reversed_numbers = numbers[::-1]

print("Original:", numbers)
print("Reversed:", reversed_numbers)
```

# Second Largest Unique Number

```python
numbers = [45, 12, 89, 34, 89, 67, 45, 23]

unique_sorted = sorted(list(set(numbers)))

second_largest = unique_sorted[-2]

print("Second-largest unique number:", second_largest)
```

# Rotate a List

```python
numbers = [10, 20, 30, 40, 50, 60]
k = 2

rotated = numbers[-k:] + numbers[:-k]

print("Rotated list:", rotated)
```

# Remove Duplicates, Preserve Order

```python
numbers = [10, 20, 10, 30, 20, 40, 30, 50]

unique_numbers = list(dict.fromkeys(numbers))

print("Without duplicates:", unique_numbers)
```

# Flatten a Nested List

```python
data = [[10, 20], [30, 40], [50, 60]]

flattened = data[0] + data[1] + data[2]

print("Flattened list:", flattened)
```

# Extract Data from a Matrix

```python
matrix = [
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
]

main_diag = [matrix[0][0], matrix[1][1], matrix[2][2]]
sum_main = sum(main_diag)

sec_diag = [matrix[0][2], matrix[1][1], matrix[2][0]]
sum_sec = sum(sec_diag)

mid_col = [matrix[0][1], matrix[1][1], matrix[2][1]]
sum_mid = sum(mid_col)

print("Main Diagonal Sum:", sum_main)
print("Secondary Diagonal Sum:", sum_sec)
print("Middle Column Sum:", sum_mid)
```
