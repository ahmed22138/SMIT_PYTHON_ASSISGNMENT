# Dictionary Access

```python
student = {"name": "Ali", "age": 22, "course": "Python", "marks": 85}

name = student["name"]
marks = student["marks"]

city = student.get("city", "Karachi")

print("Name:", name)
print("Marks:", marks)
print("City:", city)
```

# Update Dictionary

```python
employee = {"name": "Sara", "salary": 100000, "department": "IT"}

employee["salary"] = employee["salary"] + employee["salary"] * 20 // 100

employee["bonus"] = 15000

employee["department"] = "Data Engineering"

print("Updated employee:", employee)
```

# Dictionary Methods

```python
product = {"id": 101, "name": "Laptop", "price": 120000, "stock": 5}

keys = list(product.keys())
values = list(product.values())
items = list(product.items())

print("Keys:", keys)
print("Values:", values)
print("Items:", items)

stock = product.pop("stock")

print("Removed stock value:", stock)
print("Product after removal:", product)
```

# Merge Dictionaries

```python
dict1 = {"a": 10, "b": 20, "c": 30}
dict2 = {"b": 50, "d": 40}

merged = dict1 | dict2

print("Merged:", merged)
```

# Student Marks

```python
marks = {"Ali": 85, "Sara": 92, "Ahmed": 78, "Zain": 88}

highest = max(marks.values())
lowest = min(marks.values())
average = sum(marks.values()) / len(marks)

top_student = max(marks, key=marks.get)

print("Highest:", highest)
print("Lowest:", lowest)
print("Average:", average)
print("Top student:", top_student)
```

# Nested Employee Records

```python
employees = {
    "E101": {"name": "Ali", "salary": 80000},
    "E102": {"name": "Sara", "salary": 120000},
    "E103": {"name": "Ahmed", "salary": 95000}
}

sara_salary = employees["E102"]["salary"]

employees["E101"]["salary"] = employees["E101"]["salary"] + employees["E101"]["salary"] * 15 // 100

total_payroll = (employees["E101"]["salary"]
                 + employees["E102"]["salary"]
                 + employees["E103"]["salary"])

print("Sara salary:", sara_salary)
print("Ali new salary:", employees["E101"]["salary"])
print("Total payroll:", total_payroll)
```

# Dictionary Inversion

```python
data = {"Python": "Ali", "SQL": "Sara", "AWS": "Ahmed"}

inverse = dict(zip(data.values(), data.keys()))

print("Inverse dictionary:", inverse)
```

# Order Calculation

```python
order = {
    "laptop": {"price": 120000, "quantity": 2},
    "mouse": {"price": 1500, "quantity": 5},
    "keyboard": {"price": 3500, "quantity": 3}
}

subtotals = {
    "laptop": order["laptop"]["price"] * order["laptop"]["quantity"],
    "mouse": order["mouse"]["price"] * order["mouse"]["quantity"],
    "keyboard": order["keyboard"]["price"] * order["keyboard"]["quantity"]
}

grand_total = sum(subtotals.values())

discount = grand_total * 10 / 100
final_amount = grand_total - discount

print("Subtotals:", subtotals)
print("Grand total:", grand_total)
print("After 10% discount:", final_amount)
```

# Employee Performance

```python
employees = {
    "Ali": {"python": 90, "sql": 85, "aws": 88},
    "Sara": {"python": 95, "sql": 92, "aws": 90},
    "Ahmed": {"python": 80, "sql": 87, "aws": 84}
}

averages = {
    "Ali": round(sum(employees["Ali"].values()) / len(employees["Ali"]), 2),
    "Sara": round(sum(employees["Sara"].values()) / len(employees["Sara"]), 2),
    "Ahmed": round(sum(employees["Ahmed"].values()) / len(employees["Ahmed"]), 2)
}

top_performer = max(averages, key=averages.get)

highest_sql = max(employees["Ali"]["sql"], employees["Sara"]["sql"], employees["Ahmed"]["sql"])

print("Averages:", averages)
print("Top performer:", top_performer)
print("Highest SQL score:", highest_sql)
```

# Inventory Management

```python
inventory = {
    "P101": {"name": "Laptop", "price": 120000, "stock": 5},
    "P102": {"name": "Mouse", "price": 1500, "stock": 20},
    "P103": {"name": "Keyboard", "price": 3500, "stock": 10}
}

item_values = {
    inventory["P101"]["name"]: inventory["P101"]["price"] * inventory["P101"]["stock"],
    inventory["P102"]["name"]: inventory["P102"]["price"] * inventory["P102"]["stock"],
    inventory["P103"]["name"]: inventory["P103"]["price"] * inventory["P103"]["stock"]
}

total_value = sum(item_values.values())

most_valuable = max(item_values, key=item_values.get)

inventory["P101"]["stock"] -= 2
inventory["P102"]["stock"] -= 5

print("Item values:", item_values)
print("Total inventory value:", total_value)
print("Most valuable item:", most_valuable)
print("Updated inventory:", inventory)
```
