# Student Performance System

```python
students = [
    {"id": 1, "name": "Ali", "marks": (85, 90, 88)},
    {"id": 2, "name": "Sara", "marks": (92, 95, 91)},
    {"id": 3, "name": "Ahmed", "marks": (78, 84, 80)}
]

averages = {
    students[0]["name"]: round(sum(students[0]["marks"]) / len(students[0]["marks"]), 2),
    students[1]["name"]: round(sum(students[1]["marks"]) / len(students[1]["marks"]), 2),
    students[2]["name"]: round(sum(students[2]["marks"]) / len(students[2]["marks"]), 2)
}

top_student = max(averages, key=averages.get)

names = (students[0]["name"], students[1]["name"], students[2]["name"])

unique_marks = set(students[0]["marks"] + students[1]["marks"] + students[2]["marks"])

print("Averages:", averages)
print("Top student:", top_student)
print("Names:", names)
print("Unique marks:", unique_marks)
```

# Employee Skills

```python
from itertools import compress

employees = {
    "Ali": {"Python", "SQL", "AWS"},
    "Sara": {"Python", "Databricks", "SQL"},
    "Ahmed": {"AWS", "Snowflake", "SQL"}
}

common_skills = employees["Ali"] & employees["Sara"] & employees["Ahmed"]

ali_only = employees["Ali"] - employees["Sara"] - employees["Ahmed"]

all_skills = employees["Ali"] | employees["Sara"] | employees["Ahmed"]

required = {"Python", "SQL"}
meets_requirement = (
    required.issubset(employees["Ali"]),
    required.issubset(employees["Sara"]),
    required.issubset(employees["Ahmed"])
)
qualified = list(compress(employees.keys(), meets_requirement))

print("Common skills:", common_skills)
print("Ali-only skills:", ali_only)
print("All team skills:", all_skills)
print("Python + SQL employees:", qualified)
```

# Sales Data

```python
sales = [
    ("Laptop", 120000, 2),
    ("Mouse", 1500, 5),
    ("Keyboard", 3500, 3)
]

revenue_by_product = {
    sales[0][0]: sales[0][1] * sales[0][2],
    sales[1][0]: sales[1][1] * sales[1][2],
    sales[2][0]: sales[2][1] * sales[2][2]
}

total_revenue = sum(revenue_by_product.values())

top_product = max(revenue_by_product, key=revenue_by_product.get)

names, prices, quantities = zip(*sales)
price_set = set(prices)

print("Total revenue:", total_revenue)
print("Top revenue product:", top_product)
print("Name to revenue:", revenue_by_product)
print("Prices:", price_set)
```

# Customer Orders

```python
customers = {
    "C101": {"name": "Ali", "orders": [1200, 2500, 1800]},
    "C102": {"name": "Sara", "orders": [3000, 4500]},
    "C103": {"name": "Ahmed", "orders": [1500, 2000, 1000]}
}

spending = {
    customers["C101"]["name"]: sum(customers["C101"]["orders"]),
    customers["C102"]["name"]: sum(customers["C102"]["orders"]),
    customers["C103"]["name"]: sum(customers["C103"]["orders"])
}

top_customer = max(spending, key=spending.get)

ids = sorted(customers)
totals_by_id = (
    sum(customers[ids[0]]["orders"]),
    sum(customers[ids[1]]["orders"]),
    sum(customers[ids[2]]["orders"])
)

print("Spending per customer:", spending)
print("Top customer:", top_customer)
print("Totals in ID order:", totals_by_id)
```

# Course Enrollment

```python
courses = {
    "Python": {"Ali", "Sara", "Ahmed"},
    "SQL": {"Sara", "Ahmed", "Zain"},
    "AWS": {"Ali", "Ahmed", "Zain"}
}

python = courses["Python"]
sql = courses["SQL"]
aws = courses["AWS"]

all_students = python | sql | aws

all_three = python & sql & aws

at_least_two = (python & sql) | (python & aws) | (sql & aws)

exactly_two = at_least_two - all_three

exactly_one = all_students - at_least_two

unique_count = len(all_students)

print("Exactly two courses:", exactly_two)
print("All three courses:", all_three)
print("Exactly one course:", exactly_one)
print("Unique students:", unique_count)
```

# E-Commerce Transactions

```python
from itertools import compress

transactions = [
    {"id": 101, "customer": "Ali", "items": [("Laptop", 120000, 1), ("Mouse", 1500, 2)]},
    {"id": 102, "customer": "Sara", "items": [("Keyboard", 3500, 2), ("Mouse", 1500, 1)]},
    {"id": 103, "customer": "Ali", "items": [("Monitor", 45000, 1)]}
]

items_1 = transactions[0]["items"]
items_2 = transactions[1]["items"]
items_3 = transactions[2]["items"]

transaction_totals = {
    transactions[0]["id"]: items_1[0][1] * items_1[0][2] + items_1[1][1] * items_1[1][2],
    transactions[1]["id"]: items_2[0][1] * items_2[0][2] + items_2[1][1] * items_2[1][2],
    transactions[2]["id"]: items_3[0][1] * items_3[0][2]
}

total_revenue = sum(transaction_totals.values())

is_ali = (
    transactions[0]["customer"] == "Ali",
    transactions[1]["customer"] == "Ali",
    transactions[2]["customer"] == "Ali"
)
ali_spending = sum(compress(transaction_totals.values(), is_ali))

largest_transaction = max(transaction_totals, key=transaction_totals.get)

all_items = items_1 + items_2 + items_3
unique_products = set(tuple(zip(*all_items))[0])

print("Transaction totals:", transaction_totals)
print("Total revenue:", total_revenue)
print("Ali spending:", ali_spending)
print("Largest transaction ID:", largest_transaction)
print("Unique products:", unique_products)
```

# Pipeline Monitoring

```python
from itertools import compress

pipelines = {
    "pipeline_1": {"status": "success", "records": (1000, 1200, 1100)},
    "pipeline_2": {"status": "failed", "records": (800, 0, 0)},
    "pipeline_3": {"status": "success", "records": (1500, 1600, 1700)}
}

p1 = pipelines["pipeline_1"]
p2 = pipelines["pipeline_2"]
p3 = pipelines["pipeline_3"]

pipeline_totals = {
    "pipeline_1": sum(p1["records"]),
    "pipeline_2": sum(p2["records"]),
    "pipeline_3": sum(p3["records"])
}

total_records = sum(pipeline_totals.values())

is_failed = (p1["status"] == "failed", p2["status"] == "failed", p3["status"] == "failed")
failed_pipelines = list(compress(pipelines.keys(), is_failed))

largest_pipeline = max(pipeline_totals, key=pipeline_totals.get)

average_per_run = {
    "pipeline_1": round(sum(p1["records"]) / len(p1["records"]), 2),
    "pipeline_2": round(sum(p2["records"]) / len(p2["records"]), 2),
    "pipeline_3": round(sum(p3["records"]) / len(p3["records"]), 2)
}

print("Total processed records:", total_records)
print("Failed pipelines:", failed_pipelines)
print("Largest pipeline:", largest_pipeline, pipeline_totals[largest_pipeline])
print("Average records per run:", average_per_run)
```

# Banking Transactions

```python
from bisect import bisect_left

accounts = {
    "A101": {"holder": "Ali", "transactions": [5000, -2000, 3000, -1000]},
    "A102": {"holder": "Sara", "transactions": [10000, -3000, -2000]},
    "A103": {"holder": "Ahmed", "transactions": [7000, -1500, 2500]}
}

final_balances = {
    accounts["A101"]["holder"]: 0 + sum(accounts["A101"]["transactions"]),
    accounts["A102"]["holder"]: 0 + sum(accounts["A102"]["transactions"]),
    accounts["A103"]["holder"]: 0 + sum(accounts["A103"]["transactions"])
}

highest_holder = max(final_balances, key=final_balances.get)

all_transactions = sorted(
    accounts["A101"]["transactions"]
    + accounts["A102"]["transactions"]
    + accounts["A103"]["transactions"]
)
split = bisect_left(all_transactions, 0)
withdrawals = all_transactions[:split]
deposits = all_transactions[split:]

total_withdrawals = abs(sum(withdrawals))

largest_deposit = max(deposits)

print("Final balances:", final_balances)
print("Highest balance:", highest_holder, final_balances[highest_holder])
print("Total withdrawals:", total_withdrawals)
print("Largest single deposit:", largest_deposit)
```

# Product Recommendations

```python
users = {
    "Ali": {"purchased": {"Laptop", "Mouse"}, "wishlist": {"Keyboard", "Monitor", "Mouse"}},
    "Sara": {"purchased": {"Keyboard", "Monitor"}, "wishlist": {"Laptop", "Mouse"}},
    "Ahmed": {"purchased": {"Laptop", "Keyboard"}, "wishlist": {"Monitor", "Headphones"}}
}

ali_bought = users["Ali"]["purchased"]
sara_bought = users["Sara"]["purchased"]
ahmed_bought = users["Ahmed"]["purchased"]

ali_wish = users["Ali"]["wishlist"]
sara_wish = users["Sara"]["wishlist"]
ahmed_wish = users["Ahmed"]["wishlist"]

bought_by_two = (ali_bought & sara_bought) | (ali_bought & ahmed_bought) | (sara_bought & ahmed_bought)

ali_recommend = ali_wish - ali_bought

shared_wishlist = (ali_wish & sara_wish) | (ali_wish & ahmed_wish) | (sara_wish & ahmed_wish)

all_products = ali_bought | sara_bought | ahmed_bought | ali_wish | sara_wish | ahmed_wish

print("Bought by at least two users:", bought_by_two)
print("Ali wishlist minus purchases:", ali_recommend)
print("Shared wishlist products:", shared_wishlist)
print("All products:", all_products)
```

# Final Company Analysis

```python
from operator import itemgetter

company = {
    "departments": {
        "Engineering": {
            "employees": [
                ("Ali", 120000, {"Python", "AWS", "SQL"}),
                ("Sara", 150000, {"Python", "Databricks", "SQL"})
            ],
            "budget": 500000
        },
        "Analytics": {
            "employees": [
                ("Ahmed", 100000, {"SQL", "Power BI"}),
                ("Zain", 130000, {"Python", "SQL", "Tableau"})
            ],
            "budget": 400000
        }
    }
}

engineering = company["departments"]["Engineering"]
analytics = company["departments"]["Analytics"]

eng_names, eng_salaries, eng_skills = zip(*engineering["employees"])
ana_names, ana_salaries, ana_skills = zip(*analytics["employees"])

payroll = {
    "Engineering": sum(eng_salaries),
    "Analytics": sum(ana_salaries)
}

company_payroll = sum(payroll.values())

all_employees = engineering["employees"] + analytics["employees"]
highest_paid = max(all_employees, key=itemgetter(1))

shared_skills = set.intersection(*eng_skills, *ana_skills)

engineering_only = set.union(*eng_skills) - set.union(*ana_skills)

remaining_budget = {
    "Engineering": engineering["budget"] - payroll["Engineering"],
    "Analytics": analytics["budget"] - payroll["Analytics"]
}

average_salary = {
    "Engineering": payroll["Engineering"] / len(eng_salaries),
    "Analytics": payroll["Analytics"] / len(ana_salaries)
}
higher_avg_department = max(average_salary, key=average_salary.get)

name_to_salary = dict(zip(eng_names + ana_names, eng_salaries + ana_salaries))

all_unique_skills = set.union(*eng_skills, *ana_skills)

combined_budget = engineering["budget"] + analytics["budget"]
budget_sufficient = combined_budget >= company_payroll

print("1. Payroll per department:", payroll)
print("2. Company payroll:", company_payroll)
print("3. Highest-paid employee:", highest_paid[0], highest_paid[1])
print("4. Skills shared by all:", shared_skills)
print("5. Engineering-only skills:", engineering_only)
print("6. Remaining budget:", remaining_budget)
print("7. Higher average salary department:", higher_avg_department, average_salary)
print("8. Name to salary:", name_to_salary)
print("9. All unique skills:", all_unique_skills)
print("10. Combined budget sufficient:", budget_sufficient)
```
